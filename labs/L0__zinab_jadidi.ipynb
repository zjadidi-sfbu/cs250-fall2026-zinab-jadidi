## Section 2 — Why Google Colab?

### What is this?
Google Colab is a free, browser-based tool for writing and running Python code — no installation required.

### Why do we need it?
Setting up Python, Anaconda, Jupyter, and an IDE can be confusing and can differ a lot between Windows, macOS, and Chromebooks. We want you focused on **learning Python**, not troubleshooting installation problems in week one.

### One limitation to know about

The Colab **runtime** (the computer running your code behind the scenes) is **temporary**. Any files you store only in the `/content/` folder can disappear when the runtime resets (for example, after being idle for a while, or when you close and reopen the notebook).

> ⚠️ **Do not treat `/content/` as permanent storage.** Your actual notebook (the file itself) is safe once it's saved to Google Drive or GitHub — it's *other files* saved directly into `/content/` (like downloaded datasets) that can disappear.

## Section 10 — Where Does a Google Colab Notebook Actually Live?

This section is **very important** — read it carefully.

### Google Drive (your working copy)

When you use **File → Save a copy in Drive**, Colab saves the notebook into your Google Drive, usually inside a folder called **`Colab Notebooks`**.

Think of your Google Drive copy as your **working copy** — the version you edit while you're still working on an assignment.

### GitHub (your official copy)

When you use **File → Save a copy in GitHub**, Colab saves a version of your notebook into your GitHub repository.

Think of your GitHub copy as your **official, submitted copy** — this is the version your instructor will grade.

### The Colab runtime (`/content/`) — temporary

As mentioned in Section 2, files saved only in `/content/` are **temporary** and can disappear when your runtime resets. This is not where your notebook "lives" long-term.

> ⚠️ **Google Drive copy ≠ GitHub copy.** They are two separate files in two separate places. **Saving changes in Google Drive does NOT automatically update GitHub.** You must intentionally save to GitHub every time you want your GitHub copy updated.

## ⚠️ Important: Colab and GitHub Are Not Automatically Synced

### A scenario to understand

Imagine this:

1. You finish Assignment 1 and save it to GitHub using **File → Save a copy in GitHub**.
2. Later, you notice a mistake.
3. You go back to your Google Drive/Colab notebook and fix it.

**Did GitHub automatically receive the change?**

## **No.**

Fixing the notebook in Drive/Colab only updates your **Drive** copy. To update GitHub, you must go through **File → Save a copy in GitHub** **again**.

### What do I do?
1. Fix your notebook in Colab.
2. Choose **File → Save a copy in GitHub** again.
3. Use the **same repository, branch, and file path** as before, so it updates (rather than duplicates) your file.
4. Write a new, descriptive commit message, such as:
   - `Fix question 3`
   - `Complete missing question`
   - `Final Assignment 1 submission`

GitHub keeps a full history of every version you've saved — this is one of the benefits of version control! You can always look back at earlier commits if needed.

## Section 19 — Common Problems and Troubleshooting

### "I cannot see my repository from Colab."
This usually means Colab hasn't been authorized to access your GitHub account, or you're looking at the wrong GitHub account. Try **File → Save a copy in GitHub** again and complete the authorization step. Make sure you're signed into the correct GitHub account in your browser.

### "I changed my notebook but GitHub still shows the old version."
This is expected — see Section 10 and Section 14. Google Drive and GitHub are **not** automatically synced. You must save to GitHub again after making changes.

### "My notebook works when I run individual cells but fails with Run all."
This usually means cells were run out of order while you were working (for example, a variable was defined in a cell you deleted or moved). Use **Runtime → Restart session**, then **Runtime → Run all**, and fix any errors that appear — this is exactly what your instructor will see.

### "My files disappeared from /content/."
Files saved only inside `/content/` are temporary and tied to your current runtime. When the runtime resets, they're gone. Your notebook itself is safe as long as it's saved to Drive or GitHub — but other files (like downloaded datasets) are not.

### "My professor cannot open my repository."
Check two things: (1) your repository is set to **Private** (not Public — if it were public, this shouldn't be an issue, but double-check anyway) and (2) your instructor has been added as a **collaborator** and has **accepted** the invitation (Section 8).

### "I submitted the wrong GitHub link."
Go to your repository, navigate into the correct folder (`assignments/` or `labs/`), and open the exact notebook file. The URL in your browser's address bar at that point is the correct one to submit — it should end in `.ipynb` and include `/blob/main/`.

### "I accidentally made my repository public."
1. Go to your repository on GitHub.
2. Click **Settings**.
3. Scroll to the **Danger Zone** at the bottom.
4. Click **Change visibility**.
5. Select **Make private** and confirm.

### "I accidentally changed or deleted something."
GitHub keeps a history of every commit you've made. Your instructor or a tutor may be able to help you find and recover an earlier version from your commit history — don't panic, just ask for help early.
